```python
import requests
from bs4 import BeautifulSoup
import csv

# URL of the page to scrape
url = 'https://www.theoryofnext.com/next100'

r = requests.get(url)
soup = BeautifulSoup(r.content, 'html.parser')


# Check if the page was retrieved successfully
if r.status_code != 200:
    raise ValueError(f"Failed to retrieve page. Status code: {r.status_code}")
else:
    print("Page retrieved successfully.")
```

```python
# Find the section containing the ai-section, fintech-section, deeptech-section, nec-section, climate-section, healthcare-section, and saas-section 
# and check if it exists
# Check for the AI section
ai_section = soup.find('section', id='ai-section')
fintech_section = soup.find('section', id='fintech-section')
deeptech_section = soup.find('section', id='deeptech-section')
nec_section = soup.find('section', id='nec-section')
climate_section = soup.find('section', id='climate-section')
healthcare_section = soup.find('section', id='healthcare-section')
saas_section = soup.find('section', id='saas-section')

if ai_section is None:
    raise ValueError("AI section not found on the page.")
if fintech_section is None:
    raise ValueError("Fintech section not found on the page.")
if deeptech_section is None:
    raise ValueError("Deeptech section not found on the page.")
if nec_section is None:
    raise ValueError("NEC section not found on the page.")
if climate_section is None:
    raise ValueError("Climate section not found on the page.")
if healthcare_section is None:
    raise ValueError("Healthcare section not found on the page.")
if saas_section is None:
    raise ValueError("SaaS section not found on the page.")
else:
    print("All sections found successfully.")
```

```python
# in class idea_card-content-wrapper get all the divs with class hs-h5 and body-regular
idea_card_ai = ai_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_fintech = fintech_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_deeptech = deeptech_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_nec = nec_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_climate = climate_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_healthcare = healthcare_section.find_all('div', class_='idea_card-content-wrapper')
idea_card_saas = saas_section.find_all('div', class_='idea_card-content-wrapper')

# Check if any idea cards were found in the sections
idea_card = idea_card_ai + idea_card_fintech + idea_card_deeptech + idea_card_nec + idea_card_climate + idea_card_healthcare + idea_card_saas

if not idea_card:
    raise ValueError("No idea cards found in any section.")
else:
    print(f"Found {len(idea_card)} idea cards.")
```

```python
# find all the divs with class hs-h5 and body-regular under each section
def extract_idea_cards(idea_card):
    data = []
    for card in idea_card:
        title = card.find('div', class_='hs-h5').text.strip()
        description = card.find('div', class_='body-regular').text.strip()
        data.append({'title': title, 'description': description})
    return data


# Extract idea cards from each section
ai_data = extract_idea_cards(idea_card_ai)
fintech_data = extract_idea_cards(idea_card_fintech)
deeptech_data = extract_idea_cards(idea_card_deeptech)
nec_data = extract_idea_cards(idea_card_nec)
climate_data = extract_idea_cards(idea_card_climate)
healthcare_data = extract_idea_cards(idea_card_healthcare)
saas_data = extract_idea_cards(idea_card_saas)

# Combine all data into a single list
all_data = ai_data + fintech_data + deeptech_data + nec_data + climate_data + healthcare_data + saas_data

```

```python
# Check if any data was extracted
if not all_data:
    raise ValueError("No data extracted from any section.")
else:
    print(f"Extracted data from {len(all_data)} idea cards.")
```

```python
# Create a CSV file to store the data
with open('next100_ideas.csv', 'w', newline='', encoding='utf-8') as csvfile:

    # Create a CSV writer object
    writer = csv.writer(csvfile)


    fieldnames = ['Idea_card', 'Description']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    # Write the header row
    writer.writeheader()
    # Write the data rows
    for item in all_data:
        writer.writerow({'Idea_card': item['title'], 'Description': item['description']})
    print("Data written to next100_ideas.csv successfully.")

    
```
